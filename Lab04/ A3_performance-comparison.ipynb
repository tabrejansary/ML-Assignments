```python
import numpy as np
import time

def minkowski_distance(v1, v2, p=2):
    return np.sum(np.abs(v1 - v2) ** p) ** (1 / p)

def initialize_centroids(data, k):
    return data[:k].copy()

def assign_clusters(data, centroids):
    clusters = []

    for point in data:
        distances = [
            minkowski_distance(point, centroid)
            for centroid in centroids
        ]
        clusters.append(np.argmin(distances))

    return np.array(clusters)

def calculate_new_centroids(data, clusters, k):
    return np.array([
        np.mean(data[clusters == i], axis=0)
        for i in range(k)
    ])

def k_means_user(data, k, max_iterations=100):
    centroids = initialize_centroids(data, k)

    for _ in range(max_iterations):
        clusters = assign_clusters(data, centroids)
        new_centroids = calculate_new_centroids(data, clusters, k)

        if np.all(centroids == new_centroids):
            break

        centroids = new_centroids

    return clusters, centroids
```

```python
def calculate_mean(values):
    return sum(values) / len(values)

def calculate_centroid(data):
    return np.array([
        calculate_mean(data[:, i].tolist())
        for i in range(data.shape[1])
    ])

def k_means_ai(data, k, max_iterations=100):
    rng = np.random.default_rng(42)
    centroids = data[rng.choice(len(data), k, replace=False)].copy()

    for iteration in range(max_iterations):
        clusters = []

        for point in data:
            distances = [
                minkowski_distance(point, centroid)
                for centroid in centroids
            ]
            clusters.append(np.argmin(distances))

        clusters = np.array(clusters)

        new_centroids = []

        for i in range(k):
            points = data[clusters == i]

            if len(points) > 0:
                new_centroids.append(calculate_centroid(points))
            else:
                new_centroids.append(centroids[i])

        new_centroids = np.array(new_centroids)

        if np.allclose(centroids, new_centroids):
            break

        centroids = new_centroids

    return clusters, centroids
```

```python
data = np.random.default_rng(42).random((5000, 4)) * 100
k = 3

start = time.perf_counter()
k_means_user(data, k)
user_time = time.perf_counter() - start

start = time.perf_counter()
k_means_ai(data, k)
ai_time = time.perf_counter() - start

print("Performance Comparison")
print("----------------------")
print(f"Your K-Means : {user_time:.6f} seconds")
print(f"AI K-Means   : {ai_time:.6f} seconds")

if user_time < ai_time:
    print("Faster: Your version")
else:
    print("Faster: AI version")
```
