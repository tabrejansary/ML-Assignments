```python
#A2 -  unit test cases

import numpy as np

def calculate_mean(values):
    total = 0

    for value in values:
        total += value

    return total / len(values)


def minkowski_distance(vector1, vector2, p=2):
    distance = np.sum(
        np.abs(vector1 - vector2) ** p
    ) ** (1 / p)

    return distance


def calculate_centroid(cluster_points):
    centroid = []

    for column in range(cluster_points.shape[1]):
        values = cluster_points[:, column].tolist()
        centroid.append(calculate_mean(values))

    return np.array(centroid)


def k_means(data, k, max_iterations=100):
    rng = np.random.default_rng(42)

    indices = rng.choice(
        len(data),
        size=k,
        replace=False
    )

    centroids = data[indices].copy()

    for iteration in range(max_iterations):

        clusters = []

        for point in data:
            distances = []

            for centroid in centroids:
                distance = minkowski_distance(
                    point,
                    centroid,
                    2
                )

                distances.append(distance)

            closest_cluster = np.argmin(distances)
            clusters.append(closest_cluster)

        clusters = np.array(clusters)

        new_centroids = []

        for cluster_id in range(k):

            cluster_points = data[
                clusters == cluster_id
            ]

            if len(cluster_points) > 0:
                centroid = calculate_centroid(
                    cluster_points
                )
            else:
                centroid = centroids[cluster_id]

            new_centroids.append(centroid)

        new_centroids = np.array(new_centroids)

        if np.allclose(
            centroids,
            new_centroids
        ):
            centroids = new_centroids
            break

        centroids = new_centroids

    return clusters, centroids, iteration + 1

```

```python
import unittest
import numpy as np


class TestCalculateMean(unittest.TestCase):

    def test_positive_values(self):
        result = calculate_mean([10, 20, 30])
        self.assertEqual(result, 20)

    def test_negative_values(self):
        result = calculate_mean([-10, -20, -30])
        self.assertEqual(result, -20)

    def test_single_value(self):
        result = calculate_mean([50])
        self.assertEqual(result, 50)


class TestMinkowskiDistance(unittest.TestCase):

    def test_euclidean_distance(self):
        result = minkowski_distance(
            np.array([0, 0]),
            np.array([3, 4]),
            2
        )

        self.assertEqual(result, 5)

    def test_manhattan_distance(self):
        result = minkowski_distance(
            np.array([1, 2]),
            np.array([4, 6]),
            1
        )

        self.assertEqual(result, 7)

    def test_same_points(self):
        result = minkowski_distance(
            np.array([5, 10]),
            np.array([5, 10]),
            2
        )

        self.assertEqual(result, 0)


class TestCalculateCentroid(unittest.TestCase):

    def test_two_dimensional_centroid(self):
        data = np.array([
            [1, 2],
            [3, 4],
            [5, 6]
        ])

        result = calculate_centroid(data)

        expected = np.array([3, 4])

        np.testing.assert_array_equal(
            result,
            expected
        )

    def test_three_dimensional_centroid(self):
        data = np.array([
            [1, 2, 3],
            [3, 4, 5],
            [5, 6, 7]
        ])

        result = calculate_centroid(data)

        expected = np.array([3, 4, 5])

        np.testing.assert_array_equal(
            result,
            expected
        )

    def test_single_point(self):
        data = np.array([
            [10, 20]
        ])

        result = calculate_centroid(data)

        expected = np.array([10, 20])

        np.testing.assert_array_equal(
            result,
            expected
        )


class TestKMeans(unittest.TestCase):

    def test_k_means_returns_correct_shapes(self):

        data = np.array([
            [1, 1],
            [1, 2],
            [2, 1],
            [8, 8],
            [8, 9],
            [9, 8]
        ])

        clusters, centroids, iterations = k_means(
            data,
            k=2
        )

        self.assertEqual(
            len(clusters),
            len(data)
        )

        self.assertEqual(
            centroids.shape,
            (2, 2)
        )

        self.assertGreater(
            iterations,
            0
        )

    def test_k_means_creates_requested_clusters(self):

        data = np.array([
            [1, 1],
            [1, 2],
            [2, 1],
            [8, 8],
            [8, 9],
            [9, 8]
        ])

        clusters, centroids, iterations = k_means(
            data,
            k=2
        )

        unique_clusters = np.unique(clusters)

        self.assertLessEqual(
            len(unique_clusters),
            2
        )

    def test_k_means_centroids_are_valid(self):

        data = np.array([
            [1, 1],
            [1, 2],
            [2, 1],
            [8, 8],
            [8, 9],
            [9, 8]
        ])

        clusters, centroids, iterations = k_means(
            data,
            k=2
        )

        self.assertEqual(
            centroids.shape[0],
            2
        )

        self.assertEqual(
            centroids.shape[1],
            2
        )

    def test_k_means_is_deterministic(self):

        data = np.array([
            [1, 1],
            [1, 2],
            [2, 1],
            [8, 8],
            [8, 9],
            [9, 8]
        ])

        result1 = k_means(data, k=2)
        result2 = k_means(data, k=2)

        np.testing.assert_array_equal(
            result1[0],
            result2[0]
        )

        np.testing.assert_array_equal(
            result1[1],
            result2[1]
        )


unittest.main(argv=[''], exit=False)
```
